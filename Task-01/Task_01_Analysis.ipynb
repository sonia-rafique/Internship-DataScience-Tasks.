```python
# --------------- Task N0 : 1------------

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('Iris.csv')

# Display basic structure
print("Dataset Shape:", df.shape)
print("\nColumns:", df.columns)
print("\nFirst 5 Rows:\n", df.head())
print("\nLast 5 Rows:\n", df.tail())
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='SepalLengthCm',
    y='PetalLengthCm',
    hue='Species',
    style='Species',
    size='PetalWidthCm',
    sizes=(50, 200)
)

plt.title("Sepal Length vs Petal Length", fontsize=14)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Histogram to examine data distribution

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='SepalLengthCm',
    hue='Species',
    kde=True,
    bins=20
)

plt.title("Histogram of Sepal Length Distribution")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()


# Box Plot to detect outliers and spread of values

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Species',
    y='PetalLengthCm'
)

sns.stripplot(
    data=df,
    x='Species',
    y='PetalLengthCm',
    color='black',
    alpha=0.4
)

plt.title("Box Plot of Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.grid(True)
plt.show()
```

```python
# "Result: Based on the scatter plot, we observed that
# Iris-Setosa is easily distinguishable from the other
# two species due to its smaller petal dimensions.
```
